```python
class State:
    def __init__(self):
        self.monkey = "floor"
        self.box = "window"
        self.on_box = False


def monkey_banana():
    s = State()

    print("Monkey is on floor")

    print("Monkey moves to box")
    s.monkey = "window"

    print("Monkey pushes box to center")
    s.box = "center"

    print("Monkey climbs box")
    s.on_box = True

    if s.on_box and s.box == "center":
        print("Monkey touches bananas!")
        print("Goal Reached!")


monkey_banana()
```
