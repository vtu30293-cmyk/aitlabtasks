```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def inorder(root):
    if root is None:
        return
    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)

#creating tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
inorder(root)
```

```python
graph={
    'A':['B','C'],
    'B':[],
    'C':[]

}
visited=set()
def dfs(node):
  if node not in visited:
    print(node,end="")
    visited.add(node)
    for neighbour in graph[node]:
      dfs(neighbour)
print("DFS Traversal:")
dfs('A')
```

```python
from collections import deque
graph={
  'A':['B','C'],
  'B':[],
  'C':[]
}
visited=set()
queue=deque()
start='A'
queue.append(start)
print("BFS TRAVERSAL:")
while queue:
  node=queue.popleft()
  if node not in visited:
    print(node,end="")
    visited.add(node)
    for neighbour in graph[node]:
      queue.append(neighbour)
```

```python
from collections import deque
graph={
  'A':['B','C','D'],
  'B':['E','F'],
  'C':['G','H'],
  'D':['I','J'],
  'E':[],
  'F':['K','L'],
  'G':[],
  'H':[],
  'I':[],
  'J':[],
  'K':[],
  'L':[]
}
def bfs(graph,start):
    visited=set()
    queue=deque([(start,0)])
    visited.add(start)
    print("Cache invalidation error")
    while queue:
        node,time=queue.popleft()
        print(node,"_reached at",time,"second(s)")
        for neighbor in graph[node]:
          if neighbor not in visited:
            visited.add(neighbor)
            queue.append((neighbor,time+1))
bfs(graph,'A')
```
