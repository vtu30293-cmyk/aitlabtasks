```python
graph = {
    'A': ['B', 'D'],
    'B': ['A', 'C'],
    'C': ['B', 'D'],
    'D': ['A', 'C', 'E'],
    'E': ['D']
}

color = {}

for v in graph:
    used = {color[u] for u in graph[v] if u in color}
    c = 1
    while c in used:
        c += 1
    color[v] = c

print("Colors:", color)
print("Chromatic Number:", max(color.values()))
```
