```python
import heapq
def a_star_algorithm(graph, heuristics, start, goal):
    open_set = []
    heapq.heappush(open_set, (heuristics[start], start))
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0
    came_from = {}
    while open_set:
        current_f, current = heapq.heappop(open_set)
        if current == goal:
            path = [current]
            while current in came_from:
                current = came_from[current]
                path.append(current)
            return path[::-1]
        for neighbor, weight in graph.get(current, []):
            new_g = g_score[current] + weight
            if new_g < g_score[neighbor]:
                came_from[neighbor] = current
                g_score[neighbor] = new_g
                f_score = new_g + heuristics[neighbor]
                heapq.heappush(open_set, (f_score, neighbor))
    return None
my_graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('C', 1), ('D', 5)],
    'C': [('D', 1)],
    'D': []
}
my_heuristics = {
    'A': 3,
    'B': 2,
    'C': 1,
    'D': 0
}
shortest_path = a_star_algorithm(
    my_graph,
    my_heuristics,
    start='A',
    goal='D'
)
print("The optimal path found by A* is:", "".join(shortest_path))
```
