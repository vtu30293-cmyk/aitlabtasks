```python

# Ant Colony Optimization - Two Paths

pheromone1 = 2
pheromone2 = 4

alpha = 1

total = pheromone1 ** alpha + pheromone2 ** alpha

p1 = pheromone1 ** alpha / total
p2 = pheromone2 ** alpha / total

print("Path 1 Pheromone:", pheromone1)
print("Path 2 Pheromone:", pheromone2)

print("Probability of Path 1:", p1)
print("Probability of Path 2:", p2)

if p1 > p2:
    print("Ant selects Path 1")
else:
    print("Ant selects Path 2")
```
