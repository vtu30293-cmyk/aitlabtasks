```python
# Knowledge Base
employees = {
    "priya": {"department": "IT", "hobby": "dancing"},
    "divya": {"department": "IT", "hobby": "singing"},
    "kiran": {"department": "CSE", "hobby": "reading"},
    "arun": {"department": "HR", "hobby": "writing"}
}

# Query 1: Employees in IT
it_employees = tuple(
    name for name, details in employees.items()
    if details["department"] == "IT"
)

# Query 2: Employees who like reading
reading_employees = tuple(
    name for name, details in employees.items()
    if details["hobby"] == "reading"
)

# Query 3: Arun's department
arun_department = (employees["arun"]["department"].lower(),)

# Display results
print("Employees in IT:", it_employees)
print("Employees who like reading:", reading_employees)
print("Arun's department:", arun_department)
```
